```python
from config.config import load_config
config = load_config()

import psycopg2

# Достаём настройки БД
db = config.database

# Подключаемся
conn = psycopg2.connect(
    host=db.host,
    port=db.port,
    dbname=db.dbname,
    user=db.user,
    password=db.password,
)

# Проверка, что соединение живое
with conn.cursor() as cur:
    cur.execute("SELECT version();")
    print(cur.fetchone())
```
